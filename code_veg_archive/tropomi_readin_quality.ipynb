```python
import sys
import os
import numpy as np
import xarray as xr
import pandas as pd
import datetime
import fnmatch
import glob

import matplotlib as mpl
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings('ignore')
```

```python
export_path = '/Volumes/Book/HDO_L2_qual/'

for Y in range(2021,2022):
    pathlist = glob.glob(f"/Volumes/Book/HDO_L2/S5P_PAL__L2__HDO__S_{Y}06*.nc")
    #print(pathlist)
    for t,file in enumerate(pathlist):
        print(file)
        x = xr.open_datatree(file)
        #print(x)

#        for var in list(x['PRODUCT'].data_vars):
#            try:
#                x['PRODUCT'][var] = x['PRODUCT'][var].where(x['PRODUCT']['qa_value']>0.7,np.nan,drop=True)
#                print('filtered: ', var)
#            except:
#                print('could not filter: ',var)
#        print(x['PRODUCT']['deltad'])

        #print(list(x['PRODUCT']['SUPPORT_DATA']['DETAILED_RESULTS'].data_vars))
        dr = x['PRODUCT']['SUPPORT_DATA']['DETAILED_RESULTS'].to_dataset()
#        dr = dr.where(x['PRODUCT']['qa_value']>0.7,np.nan,drop=True)
        dr_dropped = dr.drop_vars(['aerosol_geometric_thickness', 'aerosol_geometric_thickness_precision', 
                                               'aerosol_height', 'aerosol_height_apriori', 'aerosol_height_precision', 
                                               'aerosol_optical_thickness', 'aerosol_optical_thickness_apriori', 
                                               'aerosol_optical_thickness_precision', 
                                               'ch4_column', 'ch4_column_apriori', 
                                               'ch4_column_prefit', 'ch4_profile_apriori', 'chi_square', 'chi_square_band', 
                                               'chi_square_prefit', 'co_column', 'co_column_precision', 'co_profile_apriori', 
                                               'convergence', 'degrees_of_freedom_total', 'degrees_of_freedom_total_prefit',
                                               'iterations', 'multiplicative_offset', 'number_of_spectral_points_in_prefit', 
                                               'number_of_spectral_points_in_retrieval', 'rms', 'spectral_shift', 'surface_albedo', 
                                               'surface_albedo_apriori', 'surface_albedo_precision', 'surface_albedo_prefit', 
                                               'surface_albedo_prefit_apriori'])
        x['PRODUCT']['SUPPORT_DATA']['DETAILED_RESULTS'] = dr_dropped
        
        #print(list(x['PRODUCT']['SUPPORT_DATA']['INPUT_DATA'].data_vars))
        ind = x['PRODUCT']['SUPPORT_DATA']['INPUT_DATA'].to_dataset()
#        ind = ind.where(x['PRODUCT']['qa_value']>0.7,np.nan,drop=True)
        ind_dropped = ind.drop_vars(['dry_air_subcolumns', 
                                     'skin_temperature', 'surface_altitude', 'surface_altitude_stdv', 
                                     'surface_pressure', 'temperature', 'u10', 'v10'])
        x['PRODUCT']['SUPPORT_DATA']['INPUT_DATA'] = ind_dropped
        
        #print(list(x['PRODUCT']['SUPPORT_DATA']['GEOLOCATIONS'].data_vars))
        geo = x['PRODUCT']['SUPPORT_DATA']['GEOLOCATIONS'].to_dataset()
#        geo = geo.where(x['PRODUCT']['qa_value']>0.7,np.nan,drop=True)
        x['PRODUCT']['SUPPORT_DATA']['GEOLOCATIONS'] = geo
        
#        print('*********************************')
#        print(x)
        x['PRODUCT']['deltad'] = x['PRODUCT']['deltad'].where(x['PRODUCT']['qa_value']>0.7,np.nan)
        x['PRODUCT']['hdo_column'] = x['PRODUCT']['hdo_column'].where(x['PRODUCT']['qa_value']>0.7,np.nan)
        x['PRODUCT']['h2o_column'] = x['PRODUCT']['h2o_column'].where(x['PRODUCT']['qa_value']>0.7,np.nan)

        export_folder = "{export_path}/{name}".format(export_path=export_path, name=file.split('/')[-1])
        x.to_netcdf(export_folder,engine='h5netcdf')
        del x
        #if t==5:
        #    sys.exit()
```
